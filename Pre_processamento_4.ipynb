```python
import pandas as pd

arquivo_oficial = "Corpus_SANTOS_TONI_FDC_tratado_sem_repeticoes.xlsx"
df = pd.read_excel(arquivo_oficial)

pd.set_option("display.max_rows", None)
candidatas = df.loc[
    df["PALAVRA"].str.contains(r"[âéêúûãõ]", case=False, na=False),
    ["PALAVRA", "LEMA", "CATMORF", "ESTSILABICA", "FREQ"]
].sort_values("FREQ", ascending=False)

print("Registros encontrados:", len(candidatas))
display(candidatas)
```

```python
# Decisões aprovadas nos blocos 1 a 4.
correcoes = {
    "melância": "melancia",
    "péssego": "pêssego",
    "colméia": "colmeia"
}

lemas = {
    "amanhã": "amanhã",
    "alguém": "alguém",
    "férias": "férias",
    "âncora": "âncora",
    "algodão": "algodão",
    "anões": "anão",
    "andréia": "andréia"
}

categorias = {
    "amanhã": "Advérbio Tempo",
    "vocês": "Pronome Pessoal Terceira Pessoa Plural",
    "inglês": "Nome Comun Masculino Singular",
    "dedão": "Nome Comun Masculino Singular",
    "picolé": "Nome Comun Masculino Singular",
    "âncora": "Nome Comun Masculino Singular",
    "algodão": "Nome Comun Masculino Singular",
    "anúncio": "Nome Comun Masculino Singular",
    "barulhão": "Nome Comun Masculino Singular",
    "cabelão": "Nome Comun Masculino Singular",
    "silêncio": "Nome Comun Masculino Singular",
    "porquê": "Nome Comun Masculino Singular",
    "português": "Nome Comun Masculino Singular",
    "cascão": "Nome Comun Masculino Singular",
    "pulão": "Nome Comun Masculino Singular",
    "barrigão": "Nome Comun Masculino Singular",
    "anões": "Nome Comun Masculino Plural",
    "tânia": "Nome Própio Pessoa",
    "célia": "Nome Própio Pessoa",
    "júlio": "Nome Própio Pessoa",
    "josé": "Nome Própio Pessoa",
    "andréia": "Nome Própio Pessoa",
    "conceição": "Nome Própio Pessoa"
}

remover = {
    "nené",
    "hércules",
    "bocão",
    "tutú",
    "peladão",
    "cocozão",
    "melecão",
    "cachorrão",
    "grúfalo",
    "porcalhão",
    "chinês",
    "joãovítor",
    "rubão",
    "botãozinho",
    "cão",
    "pernão",
    "cavalão",
    "pokémon",
    "punzão",
    "alazão",
    "joãozinho",
    "dinossaurão",
    "brunão",
}

# Decisões aprovadas no bloco 5.
lemas.update({
    "ambulância": "ambulância",
    "anéis": "anel",
    "amém": "amém",
})

categorias.update({
    "ambulância": "Nome Comun Feminino Singular",
    "anéis": "Nome Comun Masculino Plural",
    "adão": "Nome Própio Pessoa",
    "amélie": "Nome Própio Pessoa",
})

remover.update({
    "joãovitor", "tuiuiú", "ahãm", "brúfalo", "galinzé",
    "reileão", "tutúcriança", "amelié", "pões", "anaclélia",
})
```

```python
# Aplicar e salvar somente ao executar esta célula.
df = pd.read_excel(arquivo_oficial)

print("=== CORREÇÕES ENCONTRADAS ===")
for coluna in ["PALAVRA", "LEMA"]:
    for origem, destino in correcoes.items():
        quantidade = int((df[coluna] == origem).sum())
        if quantidade:
            print(f"{coluna}: {origem} -> {destino} ({quantidade} ocorrência(s))")
    df[coluna] = df[coluna].replace(correcoes)

for coluna, ajustes in [("LEMA", lemas), ("CATMORF", categorias)]:
    for palavra, valor in ajustes.items():
        mascara = df["PALAVRA"].eq(palavra) & df[coluna].ne(valor)
        if mascara.any():
            print(f"{coluna}: {palavra} -> {valor} ({int(mascara.sum())} ocorrência(s))")
            df.loc[mascara, coluna] = valor

mascara_remover = df["PALAVRA"].isin(remover)
print("=== PALAVRAS QUE SERÃO REMOVIDAS ===")
for palavra, quantidade in df.loc[mascara_remover, "PALAVRA"].value_counts().items():
    print(f"{palavra}: {quantidade}")
print("Registros que serão removidos:", int(mascara_remover.sum()))

df = df.loc[~mascara_remover].copy()
df.to_excel(arquivo_oficial, index=False)

print("Arquivo atualizado:", arquivo_oficial)
print("Total de registros restantes:", len(df))
print("Repetições restantes em PALAVRA:", int(df["PALAVRA"].duplicated().sum()))
```
