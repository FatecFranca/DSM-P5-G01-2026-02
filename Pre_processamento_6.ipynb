```python
# Revisão de inglês
import pandas as pd

arquivo = "Corpus_SANTOS_TONI_FDC_tratado_sem_repeticoes.xlsx"
base_ingles = pd.read_excel(arquivo, usecols=["PALAVRA", "LEMA"])

# Pistas de grafia usadas na triagem manual. Há falsos positivos e
# formas em inglês sem essas pistas.
pistas = r"[kwy]|sh|ck|th|ph|oo|ee|ll|tt|ff|pp|gh|wh|ng|bb|dd|mm|nn| |[bdfgjklnpqtvwxz]$"
mascara = (
    base_ingles["PALAVRA"].str.contains(pistas, case=False, na=False)
    | base_ingles["LEMA"].str.contains(pistas, case=False, na=False)
)
if "remover_ingles" not in globals():
    print("Execute a celula de decisoes para ver as exclusoes aprovadas.")
aprovadas = base_ingles["PALAVRA"].isin(globals().get("remover_ingles", set()))
preservadas = base_ingles["PALAVRA"].isin({"bacon", "jazz", "skate", "pitbul"})
triagem_ingles = base_ingles.loc[mascara | aprovadas | preservadas].copy()
triagem_ingles["DECISÃO"] = "Pista de grafia"
triagem_ingles.loc[aprovadas, "DECISÃO"] = "Remover"
triagem_ingles.loc[preservadas, "DECISÃO"] = "Manter"
print("Registros lidos na base inteira:", len(base_ingles))
display(triagem_ingles.sort_values("PALAVRA"))
```

```python
# Execute após a célula de decisões para gravar no Excel.
import pandas as pd

arquivo = "Corpus_SANTOS_TONI_FDC_tratado_sem_repeticoes.xlsx"
base = pd.read_excel(arquivo)
mascara_remover = base["PALAVRA"].isin(remover)
print("=== EXCLUSÕES ENCONTRADAS ===")
for palavra, quantidade in base.loc[mascara_remover, "PALAVRA"].value_counts().items():
    print(f"{palavra}: {quantidade}")
print("Registros que serão removidos:", int(mascara_remover.sum()))

resultado = base.loc[~mascara_remover].copy()
print("=== CORREÇÕES ENCONTRADAS ===")
total_corrigido = 0
for coluna in ["PALAVRA", "LEMA"]:
    for origem, destino in correcoes.items():
        quantidade = int(resultado[coluna].eq(origem).sum())
        if quantidade:
            print(f"{coluna}: {origem} -> {destino} ({quantidade})")
            total_corrigido += quantidade
    resultado[coluna] = resultado[coluna].replace(correcoes)

if mascara_remover.any() or total_corrigido:
    resultado.to_excel(arquivo, index=False)
    print("Arquivo atualizado:", arquivo)
else:
    print("Nenhuma alteração pendente.")

print("Total de registros restantes:", len(resultado))
print("Repetições restantes em PALAVRA:", int(resultado["PALAVRA"].duplicated().sum()))
```
