# Pré-processamento 3 - Estrutura silábica e revisão ortográfica

## Carregando a base tratada

```python
import pandas as pd
from pathlib import Path

# carrega a última versão da base sem alterar o arquivo original
df = pd.read_excel(Path("Corpus_SANTOS_TONI_FDC_tratado.xlsx"))
df.head()
```

## Verificação das estruturas silábicas

```python
# conta as estruturas silábicas mais frequentes
df['ESTSILABICA'].value_counts().head(20)
```

```python
# seleciona registros sem estrutura silábica válida
df_erros_silabicos = df[df['ESTSILABICA'] == 'erro']
df_erros_silabicos[
    ['PALAVRA', 'LEMA', 'CATMORF', 'ESTSILABICA', 'CATACENTUAL', 'FREQ']
].head(20)
```

```python
# remove registros sem estrutura silábica válida
df = df[df['ESTSILABICA'] != 'erro']
df.shape
```

## Verificação de possíveis erros ortográficos

```python
# localiza ocorrências que contêm a sequência 'voc'
df[df['PALAVRA'].str.contains('voc', case=False, na=False)][
    ['PALAVRA', 'LEMA', 'CATMORF', 'FREQ']
]
```

```python
# mostra somente as palavras exatamente iguais a 'vocé' e 'vocés'
df[df['PALAVRA'].isin(['vocé', 'vocés'])][
    ['PALAVRA', 'LEMA', 'CATMORF', 'TRANSCRICAO',
     'ACENTUACAO', 'ESTSILABICA', 'CATACENTUAL', 'FREQ']
]
```

## Remoção de palavras concatenadas e correção ortográfica

```python
# palavras dos registros identificados como concatenações
# usamos o texto para preservar os registros corretos após salvar o Excel
palavras_concatenadas = [
    'voc\u00e9est\u00e1', 'quevoc\u00e9', 'ivoc\u00e9', 'voc\u00e9vai',
    'adiantavoc\u00e9', 'amelievoc\u00e9', 'aquivoc\u00e9', 'bolinhavoc\u00e9',
    'corvocurioso', 'ivoc\u00e9deixa', 'ivoc\u00e9n\u00e3o', 'olhavoc\u00e9',
    'ondevoc\u00e9', 'poxavoc\u00e9', 'voc\u00e9quer', 'vovoc\u00e9',
    'ondecad\u00e9', 'cad\u00e9irinha'
]

# remove somente os registros concatenados selecionados
df = df[~df['PALAVRA'].isin(palavras_concatenadas)]

# corrige somente as palavras completas identificadas
df['PALAVRA'] = df['PALAVRA'].replace({
    'voc\u00e9': 'voc\u00ea',
    'voc\u00e9s': 'voc\u00eas'
})

# corrige a grafia de 'cadé' para 'cadê'
df['PALAVRA'] = df['PALAVRA'].replace({'cad\u00e9': 'cad\u00ea'})
df['LEMA'] = df['LEMA'].replace({'cad\u00e9': 'cad\u00ea'})

df['LEMA'] = df['LEMA'].replace({
    'voc\u00e9': 'voc\u00ea',
    'voc\u00e9s': 'voc\u00eas'
})

# salva a versão atualizada sem alterar a base original
df.to_excel("Corpus_SANTOS_TONI_FDC_tratado.xlsx", index=False)
df.shape
```
