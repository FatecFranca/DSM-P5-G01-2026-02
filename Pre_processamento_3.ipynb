# Pré-processamento 3 - Estrutura silábica e revisão ortográfica

## Carregando a base tratada

```python
# importando bibliotecas
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

from pandas.plotting import scatter_matrix
# importando o arquivo tratado
from pathlib import Path
df = pd.read_excel(
    Path("Corpus_SANTOS_TONI_FDC_tratado.xlsx")
)

df.head()
```

## Verificação das estruturas silábicas

```python
# conta as estruturas silábicas mais frequentes
df['ESTSILABICA'].value_counts().head(20)
```

```python
# seleciona os registros cuja estrutura silábica foi classificada como erro
df_erros_silabicos = df[df['ESTSILABICA'] == 'erro']

# mostra algumas colunas para investigar esses registros
df_erros_silabicos[
    ['PALAVRA', 'LEMA', 'CATMORF', 'ESTSILABICA', 'CATACENTUAL', 'FREQ']
].head(20)
```

```python
df_erros_silabicos.shape
```

```python
df_erros_silabicos['CATMORF'].value_counts().head(20)
```

```python
categorias_excluir = [
    'Nome Própio Pessoa',
    'Nome Própio Organização',
    'Nome Própio Lugar',
    'Interjeição'
]

df_erros_comuns = df_erros_silabicos[
    ~df_erros_silabicos['CATMORF'].isin(categorias_excluir)
]

df_erros_comuns[
    ['PALAVRA', 'LEMA', 'CATMORF', 'FREQ']
].sort_values('FREQ', ascending=False).head(30)
```

```python
df_erros_silabicos['PALAVRA'].nunique()
```

```python
# remove os registros sem estrutura silábica válida
df = df[df['ESTSILABICA'] != 'erro']

# verifica o novo tamanho da base
df.shape
```
