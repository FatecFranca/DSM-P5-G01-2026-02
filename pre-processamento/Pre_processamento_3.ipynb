## Remoção de palavras concatenadas e correção ortográfica

```python
# palavras dos registros identificados como concatenações
# usamos o texto para preservar os registros corretos após salvar o Excel
palavras_concatenadas = [
    'voc\u00e9est\u00e1', 'quevoc\u00e9', 'ivoc\u00e9', 'voc\u00e9vai',
    'adiantavoc\u00e9', 'amelievoc\u00e9', 'aquivoc\u00e9', 'bolinhavoc\u00e9',
    'corvocurioso', 'ivoc\u00e9deixa', 'ivoc\u00e9n\u00e3o', 'olhavoc\u00e9',
    'ondevoc\u00e9', 'poxavoc\u00e9', 'voc\u00e9quer', 'vovoc\u00e9',
    'ondecad\u00e9', 'cad\u00e9irinha'
]

# remove somente os registros concatenados selecionados
df = df[~df['PALAVRA'].isin(palavras_concatenadas)]

# corrige somente as palavras completas identificadas
df['PALAVRA'] = df['PALAVRA'].replace({
    'voc\u00e9': 'voc\u00ea',
    'voc\u00e9s': 'voc\u00eas'
})

# corrige a grafia de 'cadé' para 'cadê'
df['PALAVRA'] = df['PALAVRA'].replace({'cad\u00e9': 'cad\u00ea'})
df['LEMA'] = df['LEMA'].replace({'cad\u00e9': 'cad\u00ea'})

df['LEMA'] = df['LEMA'].replace({
    'voc\u00e9': 'voc\u00ea',
    'voc\u00e9s': 'voc\u00eas'
})

df.shape
```

## Limpeza das palavras de até 3 caracteres

```python
curta = df[df["PALAVRA"].astype(str).str.len().le(3)].copy()
cat = curta["CATMORF"].fillna("").astype(str)
remove_cat = cat.str.contains("Interjei|Nome Própio Outros|Nome Própio Organização", regex=True, na=False)
nomes_proprios_mantidos = {"ben", "eva", "léo", "max", "noé", "zé"}
eh_nome_proprio = cat.str.contains("Nome Própio", na=False)
remove_nome_proprio = eh_nome_proprio & ~curta["PALAVRA"].isin(nomes_proprios_mantidos)
palavras_remover = {"ah", "aha", "ahá", "ahã", "ai", "alô", "eba", "eca", "ei", "epa", "ham", "hau", "oi", "olá", "opa", "ops", "oxe", "oxi", "piu", "pum", "ufa", "uhu", "ui", "uai", "ups", "xiu", "box", "boy", "sex", "six", "toy", "zoo", "aqi", "vcé", "gfa", "gpa", "irr", "ixi", "lex", "rex", "tsu", "tsi", "vxi", "xii", "eee", "iii", "ooo", "uuu", "xxx"}
remove_manual = curta["PALAVRA"].isin(palavras_remover)
repeticao = curta["PALAVRA"].astype(str).str.len().ge(2) & curta["PALAVRA"].astype(str).str.fullmatch(r"(.)\1+", na=False)
indices_remover_curta = curta.index[remove_cat | remove_nome_proprio | remove_manual | repeticao]
df = df.drop(index=indices_remover_curta).copy()
print("Palavras curtas removidas:", len(indices_remover_curta))
print("Novo tamanho da base:", df.shape)
df.to_excel("Corpus_SANTOS_TONI_FDC_tratado_sem_repeticoes.xlsx", index=False)
```
